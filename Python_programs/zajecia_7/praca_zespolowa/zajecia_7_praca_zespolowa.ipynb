```python
import google.generativeai as genai
genai.configure(api_key="")
```

```python
def ask_agent(instruction: str, input_str: str) -> None:
  model = genai.GenerativeModel(model_name="gemini-2.5-pro", system_instruction=instruction)

  while True:
    user_input = input(input_str)

    if user_input.lower() == "exit":
      print("Do następnego!")
      break
    elif not user_input.strip():
      continue
    else:
      response = model.generate_content(user_input)
      print(response.text)
```

```python
"""Asystent do powtarzania matematycznych pojęć"""

prompt = ("Wciel się w rolę asystenta do powtarzania matematycznych pojęć, wszystko tłumacz krok po kroku, w myśl zasady dziel i zwyciężaj.")

input_message = "Wprowadź pojęcie matematyczne, które chcesz powtórzyć, lub wpisz 'exit', aby zakończyć: "
ask_agent(instruction=prompt, input_str = input_message)
```

```python
"""Asystent do wyjaśniania skomplikowanych pojęć"""

prompt = ( "Wciel się w rolę asystenta do wyjaśniania skomplikowanych pojęć, wszystko tłumacz krok po kroku, w myśl zasady dziel i zwyciężaj.")

input_message = "Wprowadź pojęcie, które chcesz zrozumieć, lub wpisz 'exit', aby zakończyć: "
ask_agent(instruction=prompt, input_str = input_message)
```
