```python
import pandas as pd
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv"
df = pd.read_csv(url)
df.head()
```

```python
print(len(df))
```

```python
print(round(df["petal_length"].mean(), 2))
```

```python
m = 0
c = ""
for spec in df.species.unique():
  averange = df[df.species == spec]["sepal_length"].mean()
  if averange > m:
    m = averange
    c = spec
print(c)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.regplot(data=df, x='petal_length', y='sepal_length', order = 3)
plt.title('Scatter plot of petal_length versus sepal_length')
plt.xlabel('Petal Length')
plt.ylabel('Sepal Length')
plt.show()
```
