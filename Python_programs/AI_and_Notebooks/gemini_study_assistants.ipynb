```python
import google.generativeai as genai
genai.configure(api_key="")
```

```python
def ask_agent(instruction: str, input_str: str) -> None:
  model = genai.GenerativeModel(model_name="gemini-2.5-pro", system_instruction=instruction)

  while True:
    user_input = input(input_str)

    if user_input.lower() == "exit":
      print("See you next time!")
      break
    elif not user_input.strip():
      continue
    else:
      response = model.generate_content(user_input)
      print(response.text)
```

```python
"""Assistant for revising mathematical concepts"""

prompt = ("Act as an assistant for revising mathematical concepts. Explain everything step by step, following the divide-and-conquer principle.")

input_message = "Enter the mathematical concept you want to revise, or type 'exit' to quit: "
ask_agent(instruction=prompt, input_str = input_message)
```

```python
"""Assistant for explaining complex concepts"""

prompt = ("Act as an assistant for explaining complex concepts. Explain everything step by step, following the divide-and-conquer principle.")

input_message = "Enter the concept you want to understand, or type 'exit' to quit: "
ask_agent(instruction=prompt, input_str = input_message)
```

```python
"""Study planning assistant"""

prompt = (
    "Act as an expert in study planning and time management. "
    "Your task is to create realistic and effective study plans. "
    "Use the divide-and-conquer principle by breaking large amounts of material into small, specific tasks. "
    "Include techniques such as Pomodoro, suggest breaks, and recommend revision methods.")

input_message = "Enter the subject, exam, or goal for which you want to create a study plan (or type 'exit' to quit): "
ask_agent(instruction=prompt, input_str = input_message)
```

```python
"""Quiz creation assistant"""

prompt = (
    "Act as a quiz expert. "
    "Your task is to create interesting and engaging quizzes on topics provided by the user. "
    "Try not to create overly simple questions and adjust their difficulty according to the user's preferences. "
)

input_message = "Enter the topic and concept of the quiz you want to create (or type 'exit' to quit): "
ask_agent(instruction=prompt, input_str = input_message)
```

```python
"""Data analysis assistant"""

prompt = (
    "Act as a data analysis assistant. "
    "Your task is to analyze data provided by the user according to the following criteria: "
    "the number of rows in the dataset, the average value of each column, and the arguments for which a given column has the highest average."
)

input_message = "Enter the data in table format (or type 'exit' to quit): "
ask_agent(instruction=prompt, input_str = input_message)
```

# Summary


*   We created five AI assistants for specific tasks defined in the first prompt. The entire process takes place remotely using API keys, communicating directly with Gemini without using a GUI.
*   Thanks to our prompts, the model knows exactly what the response should look like. User prompts are also modified to improve the quality of the responses because the model receives a more detailed question.
